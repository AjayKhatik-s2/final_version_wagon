# Wagon Eye — Phase 1 standalone wagon counter

End-to-end notebook for the **standalone** Phase-1 wagon counting package (the contents of this `wagon_count/` folder).

This notebook will:

1. Verify the runtime environment (Python, GPU, ultralytics).
2. Install dependencies from `requirements.txt`.
3. Confirm the 4 input videos and 4 `.pt` models are in place.
4. Run the full Phase-1 pipeline (per-camera tracking → master classification → cross-camera fusion → combined evidence report).
5. Inspect the final `GlobalTrainState`.
6. Inspect `results/combined_report.pdf`.

> **Phase 1 only.** No doors, no damage, no OCR, no email, no S3.
>
> Just global wagon segmentation + ENGINE/WAGON/BRAKE\_VAN classification + one combined evidence PDF.

**Output behaviour:** the pipeline does **not** store the full frame sequence of every gap any more. For each global event it keeps up to **16 representative evidence frames** — 20% / 40% / 60% / 80% of each camera's own valid evidence interval, across RIGHT_UP, LEFT_UP, RIGHT_UP_TOP and LEFT_UP_TOP — packages them into `results/combined_report.pdf`, and then deletes the temporary frames. Overlay MP4s are opt-in.

Run cells **top-to-bottom**. Each cell either prints a confirmation or raises early if something is missing.

## 5. Configure the run

Defaults match the CLI defaults of `run_global_count.py`. Edit only if you need to override a threshold.

````python
OUTPUT_DIR = os.path.join(PKG_ROOT, "results")

# Detection thresholds
SIDE_CONFIDENCE = 0.4         # YOLO conf for right_up_wagon_gap.pt + left_up_wagon_gap.pt
TOP_CONFIDENCE  = 0.4         # YOLO conf for top_gap.pt
CLASSIFICATION_SAMPLES = 5    # frames per segment voted in side_classification.pt

# Min-bbox-height / frame-height filters (camera geometry differs):
#   SIDE cameras see the gap as a TALL vertical strip -> filter at 0.35
#   TOP cameras see the gap as a THIN horizontal strip -> filter at 0.05
# If your top videos still don't show gap bboxes, lower TOP_MIN_HEIGHT_RATIO
# further (e.g. 0.02) -- the tracker's hit/miss persistence rule will still
# filter noise at the track level.
SIDE_MIN_HEIGHT_RATIO = 0.35
TOP_MIN_HEIGHT_RATIO  = 0.05

# Fusion thresholds (how strict we are about inserting a gap RIGHT_UP missed)
FUSE_MIN_SUPPORT = 2          # need >= 2 support cameras to vote in a missed gap
FUSE_MAX_SPREAD  = 1.5        # seconds: max time spread within a fusion cluster
FUSE_MIN_CONF    = 0.4        # min mean confidence to insert a fused gap

# Output toggles
# NOTE: the pipeline no longer writes one JPEG per frame per wagon.  It builds
# results/combined_report.pdf with up to 16 representative evidence frames per
# event (20/40/60/80% x 4 cameras) and deletes the temporary frames afterwards.
RENDER_VIDEOS  = False        # overlay MP4s are opt-in: ~100 MB per camera
BUILD_REPORT   = True         # build results/combined_report.pdf
REPORT_DPI     = 150          # PDF page resolution; raise for bigger evidence images
KEEP_EVIDENCE_FRAMES = False  # keep the temporary JPEGs after a successful PDF

VERBOSE = True

print(f"OUTPUT_DIR             : {OUTPUT_DIR}")
print(f"side_confidence        : {SIDE_CONFIDENCE}")
print(f"top_confidence         : {TOP_CONFIDENCE}")
print(f"side_min_height_ratio  : {SIDE_MIN_HEIGHT_RATIO}")
print(f"top_min_height_ratio   : {TOP_MIN_HEIGHT_RATIO}")
print(f"classification_samples : {CLASSIFICATION_SAMPLES}")
print(f"fuse_min_support       : {FUSE_MIN_SUPPORT}")
print(f"fuse_max_spread        : {FUSE_MAX_SPREAD}s")
print(f"fuse_min_confidence    : {FUSE_MIN_CONF}")
print(f"render_videos          : {RENDER_VIDEOS}  (opt-in)")
print(f"build_report           : {BUILD_REPORT}  -> combined_report.pdf @ {REPORT_DPI} dpi")
````

## 6. Run the full Phase-1 pipeline

Two ways to run — pick **one**:

* **6a (recommended for notebooks)** — call the pipeline functions directly. Keeps the intermediate `tracks` / `state` objects in memory so you can inspect them in later cells.
* **6b** — invoke `run_global_count.py` as a subprocess. Same outputs, but you lose the in-memory objects.

### 6a. In-process run (recommended)

````python
import time
from global_train_state import (
    GlobalTrainState, LocalCameraTracks, SegmentClass,
    MASTER_CAMERA, ALL_CAMERAS,
    CAMERA_RIGHT_UP, CAMERA_LEFT_UP, CAMERA_RIGHT_UP_TOP, CAMERA_LEFT_UP_TOP,
    summarize_state,
)
from tracker_engine import GapTracker, MasterClassifier, segments_from_gaps
import global_alignment as ga
import video_segmenter as vs
import evidence_report as er
import json

os.makedirs(OUTPUT_DIR, exist_ok=True)
PROCESSED_VIDEOS_DIR = os.path.join(OUTPUT_DIR, "processed_videos")
if RENDER_VIDEOS:
    os.makedirs(PROCESSED_VIDEOS_DIR, exist_ok=True)

# Two separate side gap models -- one per side camera.
model_paths = {
    "right_up_gap":        os.path.join(MODELS_DIR, "right_up_wagon_gap.pt"),
    "left_up_gap":         os.path.join(MODELS_DIR, "left_up_wagon_gap.pt"),
    "top_gap":             os.path.join(MODELS_DIR, "top_gap.pt"),
    "side_classification": os.path.join(MODELS_DIR, "side_classification.pt"),
}
video_paths = {
    CAMERA_RIGHT_UP:     os.path.join(INPUTS_DIR, "right_up.mp4"),
    CAMERA_LEFT_UP:      os.path.join(INPUTS_DIR, "left_up.mp4"),
    CAMERA_RIGHT_UP_TOP: os.path.join(INPUTS_DIR, "right_up_top.mp4"),
    CAMERA_LEFT_UP_TOP:  os.path.join(INPUTS_DIR, "left_up_top.mp4"),
}

# Pick the right gap model + filter settings for each camera.
def _gap_model_for(cam):
    if cam == CAMERA_RIGHT_UP:
        return model_paths["right_up_gap"]
    if cam == CAMERA_LEFT_UP:
        return model_paths["left_up_gap"]
    return model_paths["top_gap"]    # both top cameras share top_gap.pt

def _filters_for(cam):
    is_top = cam in (CAMERA_RIGHT_UP_TOP, CAMERA_LEFT_UP_TOP)
    return (
        TOP_CONFIDENCE if is_top else SIDE_CONFIDENCE,
        TOP_MIN_HEIGHT_RATIO if is_top else SIDE_MIN_HEIGHT_RATIO,
    )

t_start = time.time()

# ---- Step 1: per-camera tracking ----
print("=" * 70)
print("  STEP 1  Per-camera gap tracking")
print("=" * 70)
tracks = {}
for cam in ALL_CAMERAS:
    model = _gap_model_for(cam)
    conf, mhr = _filters_for(cam)
    tracker = GapTracker(
        camera_id=cam, model_path=model,
        confidence=conf, min_height_ratio=mhr,
        verbose=VERBOSE,
    )
    tracks[cam] = tracker.process_video(video_paths[cam], keep_raw_detections=True)

print("\nLocal counts after Step 1:")
for cam in ALL_CAMERAS:
    t = tracks[cam]
    print(f"  {cam:<14}  wagons={t.local_wagon_count:>3}  gaps={len(t.gaps):>3}  fps={t.fps:.2f}  frames={t.total_frames}")

# ---- Step 2: master classification ----
print("\n" + "=" * 70)
print("  STEP 2  RIGHT_UP master classification (ENGINE / WAGON / BRAKE_VAN)")
print("=" * 70)
master = tracks[CAMERA_RIGHT_UP]
pre_segments = segments_from_gaps(master.gaps, master.total_frames)
classifier = MasterClassifier(model_paths["side_classification"], num_samples=CLASSIFICATION_SAMPLES, verbose=VERBOSE)
initial_classifications = classifier.classify_segments(master.video_path, pre_segments)

# ---- Step 3: cross-camera fusion ----
print("\n" + "=" * 70)
print("  STEP 3  Cross-camera gap fusion")
print("=" * 70)
support = [tracks[c] for c in ALL_CAMERAS if c != CAMERA_RIGHT_UP]
fuse_cfg = dict(ga.PHASE1_DEFAULTS)
fuse_cfg.update({
    "insert_min_support": FUSE_MIN_SUPPORT,
    "insert_max_spread_sec": FUSE_MAX_SPREAD,
    "insert_min_confidence": FUSE_MIN_CONF,
})
state = ga.assemble_global_train_state(
    master_tracks=master,
    support_tracks=support,
    initial_classifications=initial_classifications,
    config=fuse_cfg,
    verbose=VERBOSE,
)

# ---- Step 4: write JSON ----
state_json_path = os.path.join(OUTPUT_DIR, "global_train_state.json")
with open(state_json_path, "w", encoding="utf-8") as f:
    f.write(state.to_json())

tracking_dump = {
    cam: tracks[cam].to_dict(include_classifications=(cam == CAMERA_RIGHT_UP))
    for cam in ALL_CAMERAS
}
if initial_classifications:
    tracking_dump[CAMERA_RIGHT_UP]["pre_fusion_classifications"] = [c.to_dict() for c in initial_classifications]
with open(os.path.join(OUTPUT_DIR, "per_camera_tracking.json"), "w", encoding="utf-8") as f:
    json.dump(tracking_dump, f, indent=2)

# ---- Step 5: overlay videos (opt-in) ----
if RENDER_VIDEOS:
    print("\n" + "=" * 70)
    print("  STEP 5  Overlay videos")
    print("=" * 70)
    for cam in ALL_CAMERAS:
        out_mp4 = os.path.join(PROCESSED_VIDEOS_DIR, f"{cam}_processed.mp4")
        try:
            vs.render_processed_video(
                local_tracks=tracks[cam],
                state=state,
                output_path=out_mp4,
                draw_raw_detections=True,
                verbose=VERBOSE,
            )
        except Exception as e:
            print(f"  ⚠ render failed for {cam}: {e}")
            state.add_note(f"render_failed:{cam}:{e}")

# ---- Step 6: combined evidence report ----
# For every global event, 4 representative frames per camera at 20/40/60/80%
# of THAT camera's own valid evidence interval -> up to 16 images per event.
# The temporary JPEGs are deleted once the PDF is written and verified.
report_info = None
if BUILD_REPORT:
    print("\n" + "=" * 70)
    print("  STEP 6  Combined evidence report (20/40/60/80% per camera)")
    print("=" * 70)
    er.warn_about_legacy_frame_dirs(OUTPUT_DIR, verbose=VERBOSE)
    try:
        report_info = er.build_combined_report(
            state=state,
            tracks=tracks,
            output_root=OUTPUT_DIR,
            dpi=REPORT_DPI,
            keep_evidence_frames=KEEP_EVIDENCE_FRAMES,
            verbose=VERBOSE,
        )
        if not report_info.get("verified"):
            state.add_note(f"report_unverified:{report_info.get('detail')}")
    except Exception as e:
        print(f"  ⚠ combined report failed: {e}")
        state.add_note(f"report_failed:{type(e).__name__}:{e}")

# Re-dump JSON with any notes from failed render/report steps
with open(state_json_path, "w", encoding="utf-8") as f:
    f.write(state.to_json())

elapsed = time.time() - t_start
print("\n" + summarize_state(state))
print(f"\n  total elapsed : {elapsed:.1f}s")
print(f"  output root   : {OUTPUT_DIR}")
if report_info:
    print(f"  report        : {report_info['pdf_path']}")
    print(f"                  {report_info['pages']} pages, "
          f"{report_info['slots_available']}/{report_info['slots_total']} evidence frames, "
          f"{report_info['detail']}")
````

## 8. Inspect the evidence report

The pipeline no longer writes per-wagon frame folders. Visual evidence lives in
`results/combined_report.pdf`: one page per global event, holding up to **16
representative frames** — 20% / 40% / 60% / 80% of each camera's own valid
evidence interval, for RIGHT_UP, LEFT_UP, RIGHT_UP_TOP and LEFT_UP_TOP.

The cell below reports the per-event evidence coverage and renders the first
event page inline so you can sanity-check that the same `GW_n` really does show
the same physical wagon across all four cameras.

````python
from IPython.display import display, HTML, Markdown

pdf_path = os.path.join(OUTPUT_DIR, er.PDF_FILENAME)

if not state.wagons:
    display(Markdown("_No wagons in state — skip._"))
else:
    # ---- per-event evidence coverage, straight from the selection logic ----
    events = er.select_event_evidence(state, tracks)
    max_slots = len(er.EVIDENCE_PERCENTAGES) * len(ALL_CAMERAS)
    print(f"Evidence coverage ({len(er.EVIDENCE_PERCENTAGES)} percentages "
          f"x {len(ALL_CAMERAS)} cameras = up to {max_slots} images per event)\n")
    header = f"  {'EVENT':<8} {'CLASS':<10} " + " ".join(f"{c:>13}" for c in ALL_CAMERAS) + "   TOTAL"
    print(header)
    print("  " + "-" * (len(header) - 2))
    for ev in events:
        cells = " ".join(f"{ev.cameras[c].available_count:>13}" for c in ALL_CAMERAS)
        print(f"  {ev.wagon.global_id:<8} {ev.wagon.classification:<10} {cells}"
              f"   {ev.available_count:>2}/{ev.total_slots}")
    total_avail = sum(e.available_count for e in events)
    total_slots = sum(e.total_slots for e in events)
    print(f"\n  TOTAL evidence frames in the report: {total_avail} / {total_slots}")

    # Any camera short of evidence? Show why (never fabricated, always explained).
    reasons = {}
    for ev in events:
        for cam in ALL_CAMERAS:
            for s in ev.cameras[cam].slots:
                if not s.available:
                    reasons[s.reason] = reasons.get(s.reason, 0) + 1
    if reasons:
        print("\n  Missing-evidence reasons:")
        for r, n in sorted(reasons.items(), key=lambda kv: -kv[1]):
            print(f"    {n:>4} x  {r}")

    # ---- embed the PDF ----
    if os.path.isfile(pdf_path):
        mb = os.path.getsize(pdf_path) / (1024 * 1024)
        display(Markdown(f"### `{er.PDF_FILENAME}` — {mb:.2f} MB"))
        rel = os.path.relpath(pdf_path, PKG_ROOT).replace(os.sep, "/")
        display(HTML(
            f'<iframe src="{rel}" width="100%" height="640" '
            f'style="border:1px solid #ccc"></iframe>'
            f'<p><a href="{rel}" target="_blank">open {er.PDF_FILENAME} in a new tab</a></p>'
        ))
    else:
        display(Markdown(
            f"_`{er.PDF_FILENAME}` not found — run cell 6a with `BUILD_REPORT = True` first._"
        ))
````

## 10. Cleanup & tips

* `results/` — all Phase-1 outputs. Safe to delete between runs. After a default
  run it holds just three files:
  * `combined_report.pdf` — the evidence deliverable (a few MB)
  * `global_train_state.json` — the canonical machine-readable contract
  * `per_camera_tracking.json` — per-camera gap timelines for debugging
* **Temporary evidence frames are removed automatically.** They are extracted to
  `results/.evidence_tmp/` and deleted as soon as the PDF is written and verified.
  They are kept only if the PDF fails verification (for diagnosis) or if you set
  `KEEP_EVIDENCE_FRAMES = True`.
* Each event carries **up to 16 evidence images**: 4 percentages
  (20/40/60/80%) × 4 camera feeds. Fewer appear only when a camera genuinely has
  no valid evidence — e.g. the event falls outside that camera's video length, or
  the interval is shorter than four frames. Those cells say
  *"No valid evidence available"* with the reason; nothing is fabricated.
* If an **earlier** run left `results/frames/` or `results/processed_videos/`
  behind, they are no longer written to and can be deleted:
  `shutil.rmtree(os.path.join(OUTPUT_DIR, "frames"), ignore_errors=True)`
* Overlay MP4s are opt-in (`RENDER_VIDEOS = True`, or `--render-videos` on the
  CLI) because they cost ~100 MB per camera.
* Phase-2 (door state / damage / OCR / loaded-empty / report) consumes
  `results/global_train_state.json`. Every `GW_n` carries its master frame range
  and time window, so a downstream stage can pull the exact frames it needs from
  the source videos with `video_segmenter.map_global_wagon_to_local_frames()` —
  the same projection the evidence report uses.
